```python
import pandas as pd

data = {
    "emp_id": range(1, 11),
    "name": [
        "Ravi", "Kiran", "Meena", "Rahul", "Priya",
        "Arun", "Sneha", "Vijay", "Anu", "Ramesh"
    ],
    "salary": [
        25000, 30000, 35000, 28000, 40000,
        32000, 45000, 27000, 38000, 33000
    ]
}

df = pd.DataFrame(data)

# Randomly select 3 employees
sample = df.sample(n=3, replace=False, random_state=42)

print(sample)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create employee data
np.random.seed(42)

df = pd.DataFrame({
    "Employee_ID": np.arange(1, 201),
    "Salary": np.random.randint(20000, 80001, size=200)
})

# 2. Randomly select 20 employees using Pandas
sample = df.sample(n=20, random_state=42)

print("Randomly Selected Employees:")
print(sample)

# 3. Visualize the selected employees
plt.figure(figsize=(10, 5))

plt.bar(
    sample["Employee_ID"].astype(str),
    sample["Salary"]
)

plt.xlabel("Employee ID")
plt.ylabel("Salary (₹)")
plt.title("Random Sample of 20 Employees - Salary")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig("probalitiysampling")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

# 1. Create employee data
np.random.seed(42)

df = pd.DataFrame({
    "Employee_ID": np.arange(1, 201),
    "Salary": np.random.randint(20000, 80001, size=200)
})

# 2. Randomly select 20 employees
sample = df.sample(n=20, random_state=42)

print("Selected Employees:")
print(sample)

# 3. Calculate statistics using SciPy
salary = sample["Salary"]

print("\nMean Salary:", stats.tmean(salary))
print("Median Salary:", np.median(salary))
print("Standard Deviation:", stats.tstd(salary))
print("Skewness:", stats.skew(salary))
print("Kurtosis:", stats.kurtosis(salary))

# 4. Visualize salaries
plt.figure(figsize=(10, 5))

plt.bar(sample["Employee_ID"].astype(str), salary)

plt.xlabel("Employee ID")
plt.ylabel("Salary (₹)")
plt.title("Random Sample: Salaries of 20 Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("NumPy + Pandas + SciPy + Matplotlib")
plt.show()
```
