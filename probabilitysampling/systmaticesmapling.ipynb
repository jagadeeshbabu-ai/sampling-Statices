```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create 1000 students
df = pd.DataFrame({
    "Student_ID": range(1, 1001),
    "Marks": np.random.randint(35, 101, size=1000)
})

# 2. Systematic sampling
np.random.seed(42)
start = np.random.randint(1, 11)

sample = df.iloc[start - 1::10]

print("Selected Students:")
print(sample)
print("Total selected:", len(sample))

# 3. Visualize selected students
plt.figure(figsize=(12, 5))

plt.scatter(
    sample["Student_ID"],
    sample["Marks"]
)

plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.title("Systematic Sampling - Student Marks")
plt.grid(True)
plt.tight_layout()
plt.savefig("systmaticsampling")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create student data
np.random.seed(42)

df = pd.DataFrame({
    "Student_ID": range(1, 1001),
    "Marks": np.random.randint(35, 101, size=1000)
})

# 2. Systematic sampling
start = np.random.randint(1, 11)
sample = df.iloc[start - 1::10]

# 3. Sort by Student ID
sample = sample.sort_values("Student_ID")

# 4. Create a clean line chart
plt.figure(figsize=(12, 5))

plt.plot(
    sample["Student_ID"],
    sample["Marks"],
    marker="o",
    markersize=4,
    linewidth=1.5,
    label="Student Marks"
)

plt.title("Student Marks - Systematic Sampling")
plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.xticks(sample["Student_ID"][::5], rotation=45)
plt.ylim(30, 105)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.legend()
plt.tight_layout()
# plt.savefig("2")
plt.show()
```
