```python
import pandas as pd

# Data about banking professionals
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena", "Rahul"],
    "Role": [
        "Fraud Analyst",
        "Data Scientist",
        "Bank Manager",
        "Machine Learning Engineer",
        "Student"
    ],
    "Experience": [8, 5, 12, 6, 0]
})

# Select people suitable for fraud detection research
sample = df[
    (df["Experience"] >= 5) &
    (df["Role"] != "Student")
]

print("Judgment Sampling Results:")
print(sample)
```
