```python
import numpy as np
import pandas as pd

np.random.seed(42)

df = pd.DataFrame({
    "Student_ID": range(1, 101),
    "Gender": np.random.choice(["Male", "Female"], 100)
})

# Quota sampling: select 5 from each group
male_sample = df[df["Gender"] == "Male"].head(5)
female_sample = df[df["Gender"] == "Female"].head(5)

sample = pd.concat([male_sample, female_sample])

print(sample)
print(sample["Gender"].value_counts())
```
