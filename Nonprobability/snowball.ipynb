```python
import pandas as pd

# Initial participants
participants = {
    "Name": ["Ravi", "Sita", "Arun", "Priya", "Kiran"]
}

df = pd.DataFrame(participants)

# Snowball sampling: participants refer new participants
referrals = {
    "Ravi": ["Rahul", "Meena"],
    "Sita": ["Anu"],
    "Arun": ["Vijay", "Rani"]
}

# Collect referred participants
new_participants = []

for people in referrals.values():
    new_participants.extend(people)

new_df = pd.DataFrame({
    "Name": new_participants
})

# Combine initial and referred participants
sample = pd.concat([df, new_df], ignore_index=True)

print(sample)
```
